```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
from scipy.io import wavfile
plt.rcParams["figure.figsize"] = (9, 3.5)
```

```python
n = np.arange(-5, 11)
delta = (n == 0).astype(float) # unit impulse
step = (n >= 0).astype(float) # unit step
ramp = n * step # unit ramp
expo = 0.8**n * step # real exponential
fig, axes = plt.subplots(1, 4, figsize=(13, 2.6))
for ax, x, title in zip(axes, [delta, step, ramp, expo],
 ["δ(n)", "u(n)", "r(n)", "(0.8)^n u(n)"]):
 ax.stem(n, x, basefmt="k-")
 ax.set_title(title); ax.set_xlabel("n")
plt.tight_layout(); plt.show()
```

```python
n = np.arange(-5, 11)
step = (n >= 0).astype(float)

a_values = [0.5, -0.8, 1.1]

fig, axes = plt.subplots(1, 3, figsize=(13, 4))

for ax, a in zip(axes, a_values):
    x = a**n * step

    ax.stem(n, x)
    ax.set_title(f"a = {a}")
    ax.set_xlabel("Sample index (n)")
    ax.set_ylabel("Amplitude")
    ax.grid(True)

plt.tight_layout()
plt.show()
```

```python
n = np.arange(0, 41)

x1 = np.cos(2 * np.pi * n / 8)
x2 = np.cos(2 * np.pi * n / 7.5)

fig, axes = plt.subplots(2, 1, figsize=(12, 6))

axes[0].stem(n, x1)
axes[0].set_title("Cosine Signal: cos(2πn/8)")
axes[0].set_xlabel("Sample index (n)")
axes[0].set_ylabel("Amplitude")
axes[0].grid(True)

axes[1].stem(n, x2)
axes[1].set_title("Cosine Signal: cos(2πn/7.5)")
axes[1].set_xlabel("Sample index (n)")
axes[1].set_ylabel("Amplitude")
axes[1].grid(True)

plt.tight_layout()
plt.show()
```

```python
import os
from IPython.display import Audio

fname = "my_voice.wav"

if os.path.exists(fname):

    fs, x = wavfile.read(fname)

    x = x.astype(float)

    # Convert stereo to mono
    if x.ndim > 1:
        x = x.mean(axis=1)

    # Normalize the amplitude
    peak = np.max(np.abs(x))

    if peak > 0:
        x = x / peak

    print("Using my voice recording")

else:

    # Generate a fallback test signal
    fs = 16000
    t = np.arange(0, 2, 1/fs)

    x = 0.5 * np.sin(2*np.pi*440*t) * np.exp(-t)

    print("Using fallback test signal")

t = np.arange(len(x)) / fs

print("Sampling rate:", fs, "Hz")
print("Duration:", len(x)/fs, "seconds")
print("Number of samples:", len(x))

plt.figure(figsize=(12, 4))
plt.plot(t, x)
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.title("Audio Signal Waveform")
plt.grid(True)
plt.show()
```
